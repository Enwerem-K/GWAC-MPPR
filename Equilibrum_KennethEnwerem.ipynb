```python
from sympy import symbols, Eq, solve
```

```python
# Define symbols
S_h, I_h, R_h, S_v, I_v, N_h, N_v = symbols("S_h I_h R_h S_v I_v N_h N_v")
Λ_h, ω_h, θ, a, b, μ_h, σ, γ_h, δ_h = symbols('Λ_h ω_h θ a b μ_h σ γ_h δ_h')
Λ_v, ν, η, c, τ, μ_v = symbols('Λ_v ν η c τ μ_v')
```

```python
# Define the endemic equations
eq1 = Eq(Λ_h + ω_h*R_h - (1 - θ)*a*b*I_v*S_h/N_h - μ_h*S_h, 0)
eq2 = Eq((1 - θ)*a*b*I_v*S_h/N_h - (σ + γ_h + δ_h + μ_h)*I_h, 0)
eq3 = Eq((σ + γ_h)*I_h - (ω_h + μ_h)*R_h, 0)
eq4 = Eq((1 - ν - η)*Λ_v - (1 - θ)*a*c*I_h*S_v/N_h - (η + θ + τ + μ_v)*S_v, 0)
eq5 = Eq((1 - θ)*a*c*I_h*S_v/N_h - (η + θ + τ + μ_v)*I_v, 0)
```

```python
# Solve the system of equations for endemic equilibrium
solution = solve([eq1, eq2, eq3, eq4, eq5], (S_h, I_h, R_h, S_v, I_v))
solution
```
